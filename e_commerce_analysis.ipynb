```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("ecommerce_data.csv")
```

```python
df.drop_duplicates('Order_ID')
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'],errors='coerce')
```

```python
df['Order_Date'].dtype
```

```python
df['Payment_Method'] = df['Payment_Method'].fillna('Unknown')

df['Customer_Rating'] = df['Customer_Rating'].fillna(df['Customer_Rating'].median())

df['Delivery_Days'] = df['Delivery_Days'].fillna(df['Delivery_Days'].median())
```

```python
df.isnull().sum()
```

```python
# EDA
 # Total Sales

total_sales = df['Sales'].sum()
print(total_sales)
```

```python
# Total Profit

total_profit = df['Profit'].sum()
print(total_profit)
```

```python
# Total Orders

toatal_orders = df['Order_ID'].count()
print(toatal_orders)

# OR 

total_orders = df["Order_ID"].nunique()
print(total_orders)
```

```python
total_customers = df['Customer_ID'].count()
print(total_customers)
```

```python
# Average Order Value

average_order_value = df['Sales'].sum() / df['Order_ID'].count()
print(average_order_value)

  # OR

average_order_value = df["Sales"].sum() / df["Order_ID"].nunique()
print(average_order_value)
```

```python
# Category Analysis
# Which category generates the most revenue ?

category_analysis = (df.groupby('Category').agg(
          Sales=('Sales','sum'),
          Profit=('Profit','sum'),
          Orders=('Order_ID','nunique'),
          Quantity=('Quantity','sum')).sort_values("Sales", ascending=False))

print(category_analysis)
```

```python
# State Analysis

state_analysis = (df.groupby('State').agg(
    Sales=('Sales','sum'),
    Profit=('Profit','sum'),
    Orders=('Order_ID','nunique')).sort_values('Sales',ascending=False))

print(state_analysis)
```

```python
# Monthly Trend

monthly_sales = (df.groupby(df["Order_Date"].dt.to_period("M")).agg(
          Sales=("Sales","sum"),
          Profit=("Profit","sum")))

print(monthly_sales)
```
